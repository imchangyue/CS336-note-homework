# scaling laws
**核心思想**
在过去，训练大型语言模型是一个繁琐且效率低下的过程。研究者们通常在大型模型上反复调整超参数（hyperparameters），比如学习率、批次大小等，来寻找最佳性能。这个过程非常耗时且成本高昂，正如幻灯片中提到的 "Old and unpleasant"（老旧且令人不快）。

而新的方法（"New (over?) optimism"）则更加高效。研究发现，**大型语言模型的性能（如验证集损失）与其规模（如参数量、计算量、数据集大小）之间存在简单且可预测的数学关系**。这意味着：

 - 我们可以在小型模型上进行超参数调优，找到最佳配置。
 - 然后，我们可以利用这些缩放法则，将调优结果外推（extrapolate）到大型模型，从而预测大型模型在最佳配置下的性能。

#### 数据量对模型训练的影响

 - 在数据量少时，性能提升有限。
 - 在中间的“黄金”区域（幂律区域），数据量和模型性能之间存在可预测的、强有力的关系，这是我们最应该关注的区域。
 - 当数据量达到一定程度后，模型性能会达到一个天花板，即不可约误差。此时，继续增加数据可能不是最优的策略。

### 应用


进行预算规划：在确定了总的计算资源（成本）后，我们可以通过这个公式来决定是应该把资源更多地投入到数据收集，还是模型参数的增加上，以实现最低的损失。

避免浪费：正如我们之前讨论的，如果盲目地增加数据或模型而没有平衡，可能会导致资源浪费。缩放法则可以帮助我们找到一个“甜点”，即数据和模型大小的最佳组合。

#### 在扩展模型时，同时增加数据量与模型大小同样重要



### 问题
在实际部署中，大部分的计算成本都用于模型推理（Inference）。这意味着，一个模型一旦训练完成并被投入使用，其推理产生的计算量将远远超过一次性训练所消耗的计算量。因此，幻灯片提出，我们应该“过度训练”（‘over’ train）模型。

tokens/param 比率。这个比率非常关键，它代表了每个模型参数在训练时“看到”了多少个数据 tokens。

GPT-3: 2 tokens / param

Chinchilla: 20 tokens / param

LLaMA 65B: 22 tokens / param

Llama 2 70B: 29 tokens / param

Mistral 7B: 110 tokens / param

Llama 3 70B: 215 tokens / param

这个列表清晰地展示了一个趋势：越新的模型，其 tokens/param 比率越高。

早期的 GPT-3 参数量巨大，但训练数据量相对较少，导致每个参数的训练经验不足。

Chinchilla 论文纠正了这一问题，指出更合理的比率应该更高。

后续的 Llama 系列和 Mistral 模型，都遵循了这一原则，显著提高了 tokens/param 比率。特别是 Mistral 7B，作为一个相对小规模的模型，它通过极高的 tokens/param 比率，在性能上可以媲美甚至超越一些更大的模型。Llama 3 70B 则将这一比率推向了新的高度

#### 又一

缩放法则不仅仅适用于语言模型，它也普遍存在于其他类型的模型中，例如扩散模型。

不同类型的模型具有不同的缩放特性。在特定的计算预算下，一种模型可能比另一种模型更有效，但当计算量足够大时，这种关系可能会逆转。
